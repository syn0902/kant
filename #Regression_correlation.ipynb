**상관분석, 회귀분석**


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

from sklearn.linear_model import LinearRegression
```

```python
#### 상관계수 값과 시각화 비교


# 데이터 (예제 데이터)
X = [-2, -1, 0, 1, 2]
Y = [4, 1, 0, 1, 4]  # Y = X²

# 상관계수 계산
pearson_r, _ = stats.pearsonr(X, Y)
spearman_r, _ = stats.spearmanr(X, Y)
kendall_r, _ = stats.kendalltau(X, Y)

print(f"피어슨 상관계수 (Pearson): {pearson_r:.2f}")
print(f"스피어만 상관계수 (Spearman): {spearman_r:.2f}")
print(f"켄달 상관계수 (Kendall): {kendall_r:.2f}")
```

```python
## 시각화 - 산점도
plt.figure(figsize=(6, 4))

plt.scatter(X, Y, color='royalblue', s=80, zorder=5)

plt.title('Scatter Plot of Y = X² (Correlation = 0)')
plt.xlabel('X')
plt.ylabel('Y')
plt.ylim(-0.5, 5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

```python
# 예제 데이터 - 공부 시간과 점수와의 관계

X = np.array([2, 3, 4, 5, 7])  # 공부 시간
Y = np.array([5, 7, 9, 10, 13])  # 시험 점수
```

```python
# Y.mean()
```

```python
import matplotlib.pyplot as plt


# 산점도 그리기
plt.scatter(X, Y, color='blue', label='Study time vs Test score')

# 그래프 타이틀 및 축 레이블 추가
plt.title('Scatter Plot: Study time vs Test score')
plt.xlabel('Study time (hours)')
plt.ylabel('Test score')

# 그래프에 레전드 추가
plt.legend()

# 그래프 표시
plt.show()
```

```python
# 상관계수 계산
correlation = np.corrcoef(X, Y)
correlation_coefficient = correlation[0, 1] #  0번째 행, 1번째 열(Y) 값

print("상관계수:", correlation_coefficient)

# numpy 라이브러리의 np.corrcoef(X, Y) 함수, 피어슨(Pearson) 상관계수 계산
# cf, scipy.stats.pearsonr
```

```python
print(correlation[0, 0]) # X와 X의 관계
print(correlation[0, 1]) # X와 Y의 관계
print(correlation[1, 0]) # Y와 X의 관계
print(correlation[1, 1]) # Y와 Y의 관계
```

```python
# 선형 회귀 모델 생성 및 훈련
model = LinearRegression()  # 선형 회귀 모델 객체 생성
model.fit(X.reshape(-1, 1), Y) # 모델 학습, Y:정답 데이터

# 회귀계수
regression_coefficient = model.coef_[0] # 변수가 여러 개일 수 있기 때문에 배열 형태로 저장
print("회귀계수:", regression_coefficient)

# 결정계수 (R²) 계산
r_squared = model.score(X.reshape(-1, 1), Y)
print("결정계수 (R²):", r_squared)
```

```python
# 산점도 그리기
plt.scatter(X, Y, color='blue', label='Study time vs Test score')

# 그래프 타이틀 및 축 레이블 추가
plt.title('Scatter Plot: Study time vs Test score')
plt.xlabel('Study time (hours)')
plt.ylabel('Test score')

# 회귀선 그리기
# 학습 모델(model)에 입력 데이터 X 입력 -> 모델이 계산한 예측 결과값을 Y_pred 변수에 저장
Y_pred = model.predict(X.reshape(-1, 1))  # 예측 값 저장
plt.plot(X, Y_pred, color='red', label='Regression line')

# 그래프에 레전드 추가
plt.legend()


# 파란색 점 : 실제 데이터, 빨간색 직선 : 예측값
# 선형 회귀 모델이 현실 데이터를 얼마나 잘 근사하는지 시각적으로 확인
# 점과 선 사이의 거리 = 오차 확인
```
